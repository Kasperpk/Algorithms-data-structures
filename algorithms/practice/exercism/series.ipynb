## Problem statement

In this problem i am told work with a series of numerical digits. This is a string of digits such as '49142' which does not have to be *numerically* consecutive. 

What the program needs to do is output all contigious substrings of some length `n` provided as paramater `length` to the function. 

So for the inputs `49142` and `3` there would be three such outputs. 

* "491"
* "914" 
* "142"

Notice that we do contain overlaps in the outputs as we only move one position along the series and try to find the next contigious substring of length `3` in this case. 

### The plan

First of all i need to deal with cases where 

1. The length > series
2. The length is 0
3. The length is < 0
4. The series is empty

The plan here is to scan over the series looking to find a substring starting at the first position. 

That is something like series[cur_index:n]

Where i then increment cur_index. What the program should check is if cur_index + n is greater than the length of the series because that the program has to halt. For example with the number `49142`, we start from cur_index = 0. 

Then we grab series[cur_index:cur_index+(n-1)] which 's '491'. Here we do n-1 because strings are 0 indexes in python. Then we increment cur_index so we grab again series[cur_index:cur_index+(n-1)] which is '914'. Incrementing cur_index again now at 2 we have series[cur_index:cur_index+(n-1)] where we get from index 2:4 whic is '142'. Incrementing it again we have cur_index = 3. The last index of this series is 4 and so cur_index+(n-1) > 4 which means we should stop and return what we have. 

```python
def slices(series, length):
    '''
    Function to output slices of consecutive characters of certain length from series

    Arguments:
        series: Input series of type string
        length: Length of consecutive characteres
    '''
    
    if length == 0:
        # if the length of slice is zero. 
        raise(ValueError("slice length cannot be zero"))

    if not series:
        # if the series provided is empty.
        raise ValueError("series cannot be empty")
    
    if length < 0:
        # if the length of slice is negative.
        raise(ValueError("slice length cannot be negative"))
    

    if length > len(series):
        # if the length of slice is greater than series length.
        raise(ValueError("slice length cannot be greater than " \
        "series length"))
    
    
    cur_index = 0

    slices = []

    while cur_index + (length) <= len(series):
        slices.append(series[cur_index:cur_index+(length)])
        cur_index += 1

    return slices
```
