Given a number, your task is to express it in English words exactly as your friend should say it out loud. Yaʻqūb expects to use numbers from 0 up to 999,999,999,999.

Examples:

* 0 → zero
* 1 → one
* 12 → twelve
* 123 → one hundred twenty-three
* 1,234 → one thousand two hundred thirty-four

so in this case we both have to look at the digit and it's position. Which can range from the ones place up to hundreds of millions. So the last number 999,999,999 in wording is Nine hundred ninety-nine million, nine hundred ninety-nine thousand, nine hundred ninety-nine. 

Okay so i think i have to sort of loop through the number in a reverse order checking the value of the digit and using the index to determine it's place value. 

```python
WRITTEN_FORM_SPECIAL_CASES = {10: 'Ten', 11: 'Eleven', 12: 'twelve', 13:'thirdteen', 14: 'fourteen', 15: 'fifteen', 16: 'sixteen', 17: 'seventeen', 18: 'eighteen', 19: 'nineteen'}

PLACE_VALUES = {
    'Ones': {1: 'one', 2: 'two', 3: 'three', 4: 'four', 5: 'five', 6: 'six', 7: 'seven', 8: 'eight', 9: 'nine'},
    'Tens': {2: 'twenty', 3: 'thirty', 4: 'fourty', 5: 'fifty', 6: 'sixty', 7: 'seventy', 8: 'eighty', 9: 'ninety'}}
```

```python
def written_form_number(number):

    num_len = len(str(number))
    num_string = str(number)
    written_form = ''

    if number == 0:
        return 'zero'

    if number < 0 or number > 1000000000000:
        raise ValueError("input out of range")

    while num_len > 0:

        if int(num_string) in WRITTEN_FORM_SPECIAL_CASES.keys():
            if len(written_form) == 0:
                written_form = written_form + WRITTEN_FORM_SPECIAL_CASES[number]
            else:
                written_form = written_form + ' ' + WRITTEN_FORM_SPECIAL_CASES[number]
                
            return written_form
        
        if num_len == 12:

            # if len num is 10 then the written form adds 1-9 billion. 
            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred billion'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred billion'
            
            num_string = num_string[1:]
            num_len = num_len - 1
        
        if num_len == 11:

            # if len num is 10 then the written form adds 1-9 billion. 
            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Tens'][int(num_string[0])] 
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Tens'][int(num_string[0])]
            
            num_string = num_string[1:]
            num_len = num_len - 1
        
        if num_len == 10:

            # if len num is 10 then the written form adds 1-9 billion. 
            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'billion'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'billion'
            
            num_string = num_string[1:]
            num_len = num_len - 1
        
        if num_len == 9:

            # if len num is 9 then the written form adds 1-9 hundred million. 

            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
                
            num_string = num_string[1:]
            num_len = num_len - 1
        
        
        if num_len == 8:

            # if len num is 8 then the written form adds 1-9 thousands million
            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Tens'][int(num_string[0])] 
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Tens'][int(num_string[0])]
            
            num_string = num_string[1:]
            num_len = num_len - 1
        
        
        if num_len == 7:

            # if len num is 7 the written form adds 10-90 million

            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'million'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'million'
            
            
            num_string = num_string[1:]
            num_len = num_len - 1
    
        if num_len == 6:

            # if len num is 6 the written form adds 1-9 million

            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
            
            num_string = num_string[1:]
            num_len = num_len - 1

        if num_len == 5:

            if int(num_string[0]) != 0:
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Tens'][int(num_string[0])] 
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Tens'][int(num_string[0])]
            
            num_string = num_string[1:]
            num_len = num_len - 1
        
        if num_len == 4:

            if int(num_string[0]) != 0:
               
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'thousand'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'thousand'
        
            num_string = num_string[1:]
            num_len = num_len - 1

        if num_len == 3:

            if int(num_string[0]) != 0:

                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
                else:
                    written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(num_string[0])] + ' ' + 'hundred'
            
            num_string = num_string[1:]
            num_len = num_len - 1

        if num_len == 2:
    
            if int(num_string[0]) != 0:

                if int(num_string) in WRITTEN_FORM_SPECIAL_CASES.keys():
                    if len(written_form) == 0:
                        written_form = written_form + WRITTEN_FORM_SPECIAL_CASES[int(num_string)]
                    else:
                        written_form = written_form + ' ' + WRITTEN_FORM_SPECIAL_CASES[int(num_string)]
                    return written_form
    
                else:
                    if len(written_form) == 0:
                        written_form = written_form + PLACE_VALUES['Tens'][int(num_string[0])]
                    else:
                        written_form = written_form + ' ' + PLACE_VALUES['Tens'][int(num_string[0])]
            
            num_string = num_string[1:]
            num_len = num_len - 1

        if num_len == 1:
            
            if int(num_string[0]) != 0:
            
                if len(written_form) == 0:
                    written_form = written_form + PLACE_VALUES['Ones'][int(num_string[0])]
                else:
                    written_form = written_form + '-' + PLACE_VALUES['Ones'][int(num_string[0])]
            num_string = num_string[1:]
            num_len = num_len -1
    
    return written_form


    

```
