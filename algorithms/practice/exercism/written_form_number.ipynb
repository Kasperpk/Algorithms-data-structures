Given a number, your task is to express it in English words exactly as your friend should say it out loud. Yaʻqūb expects to use numbers from 0 up to 999,999,999,999.

Examples:

* 0 → zero
* 1 → one
* 12 → twelve
* 123 → one hundred twenty-three
* 1,234 → one thousand two hundred thirty-four

so in this case we both have to look at the digit and it's position. Which can range from the ones place up to hundreds of millions. So the last number 999,999,999 in wording is Nine hundred ninety-nine million, nine hundred ninety-nine thousand, nine hundred ninety-nine. 

Okay so i think i have to sort of loop through the number in a reverse order checking the value of the digit and using the index to determine it's place value. Another importan thing to know is that large numbers are broken into groups of three digits (called periods). You only say the name of the big category (like billion, million, or thousand) at the very end of each three-digit group. Inside each group, you just read it like a normal, everyday number under 1,000.

That means i have to identify the groups in the number. I can use the length of the number. I can do a split on a converted string. 

```python
WRITTEN_FORM_SPECIAL_CASES = {10: 'ten', 11: 'eleven', 12: 'twelve', 13:'thirdteen', 14: 'fourteen', 15: 'fifteen', 16: 'sixteen', 17: 'seventeen', 18: 'eighteen', 19: 'nineteen'}

PLACE_VALUES = {
    'Ones': {1: 'one', 2: 'two', 3: 'three', 4: 'four', 5: 'five', 6: 'six', 7: 'seven', 8: 'eight', 9: 'nine'},
    'Tens': {2: 'twenty', 3: 'thirty', 4: 'forty', 5: 'fifty', 6: 'sixty', 7: 'seventy', 8: 'eighty', 9: 'ninety'}}
```

```python
def written_form_number(number):

    if number < 0 or number > 100000000000:
        raise ValueError("input out of range")

    if number == 0:
        return 'zero'

    num_string = str(number)

    number_values = {'billion': {12: '', 11: '', 10: ''},
                     'million': {9: '', 8: '', 7: ''},
                     'thousand': {6: '', 5: '', 4: ''},
                     'hundred': {3: '', 2: '', 1: ''}}
    

    num_index = 0

    for position in range(len(num_string), 0, -1):
        
        if position > 9:
            number_values['billion'][position] = num_string[num_index]
            num_index = num_index + 1
        elif position > 6:
            number_values['million'][position] = num_string[num_index]
            num_index = num_index + 1
        elif position > 3:
        
            number_values['thousand'][position] = num_string[num_index]
            num_index = num_index + 1
        elif position == 3:
            number_values['hundred'][position] = num_string[num_index]
            num_index = num_index + 1
        elif position == 2:
            number_values['hundred'][position] = num_string[num_index]
            num_index = num_index + 1
        else:
            number_values['hundred'][position] = num_string[num_index]
            num_index = num_index + 1


    written_form = ''

    for place, value in number_values.items():
        for position, val in value.items():
            if val and val != '0':
                if position in [12,9,6,3]:
                    if len(written_form) == 0:
                            written_form = written_form + PLACE_VALUES['Ones'][int(val)] + ' ' + 'hundred' 
                    else:
                        written_form = written_form + ' ' +  PLACE_VALUES['Ones'][int(val)] + ' ' + 'hundred' 

                if position in [11,8,5,2]:
                    if int(str(val) + value[position-1]) in WRITTEN_FORM_SPECIAL_CASES.keys():
                        if len(written_form) == 0:
                            written_form = written_form + WRITTEN_FORM_SPECIAL_CASES[int(str(val) + value[position-1])]
                        else:
                            written_form = written_form + ' ' + WRITTEN_FORM_SPECIAL_CASES[int(str(val) + value[position-1])]
                        break

                    elif len(written_form) == 0:
                        written_form = written_form + PLACE_VALUES['Tens'][int(val)] 
                    else:
                        written_form = written_form + ' ' + PLACE_VALUES['Tens'][int(val)]

                if position in [10, 7, 4, 1]:
        
                    if len(written_form) == 0:
                            written_form = written_form + PLACE_VALUES['Ones'][int(val)]

                    

                    elif value[position+1] != '0':
                        
                        written_form = written_form + '-' + PLACE_VALUES['Ones'][int(val)]
                    else:
                        written_form = written_form + ' ' + PLACE_VALUES['Ones'][int(val)]

        

        if written_form != '' and place == 'billion' and set(value.values()) != set('0'):
            written_form = written_form + ' ' + place

        elif written_form != '' and place == 'million' and set(value.values()) != set('0'):
            written_form = written_form + ' ' + place

        elif written_form != '' and place == 'thousand' and set(value.values()) != set('0'):
            written_form = written_form + ' ' + place
        
        
    return written_form
```
