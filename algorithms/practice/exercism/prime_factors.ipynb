Compute the prime factors of a given natural number also known as prime factorization. 

To make it clear a prime number is only evenly divisible by itself and 1. Note that 1 is not a prime number.

One algorithm is called the Factor Tree Method. It starts with the original number and finds two numbers which product yields that number. Then from there finding two numbers which product yields that new number and so on until we have prime numbers which are the basic building blocks of that original number.

Here is an example of using the Factor Tree Method. We end up with the prime factorization of $2*2*2*3*5$ which are the basic building blocks of the number 120. 



So now that we better understand the problem, how can we devise a plan for the algorithm. First we need to find two numbers that make up the original number until that number is a prime number itself. When it is then we add it to the list of prime factors. We know it's a prime number when we can't divide it evenly with any other number other than 1. For example 11 cannot be divided and return a whole integer number so if that was the original number we would have to just add that to the prime factorization and return.

```python
import sympy

def prime_factors(value):

    # 1 is not a prime
    if value == 1:
        return []
    
    prime_factor_list = []

    prime = 2

    while value != 1:

        if value%prime == 0:
            value = int(value/prime)
            prime_factor_list.append(prime)
        else:    
            prime = sympy.nextprime(prime)

    return prime_factor_list
```
