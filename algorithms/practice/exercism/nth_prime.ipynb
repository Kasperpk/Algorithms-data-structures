```python
def nth_prime(number):

    if number == 0:
        ValueError('There is no zeroth prime')

    list_of_primes = [2]
    num = 2
    while len(list_of_primes) != number:
        print(f"Currently the follwing primes have been found: {list_of_primes}")
        print(f"Currently testing if {num} is a prime by checking if it is equally \
              divisible by any of the primes")
        lp_index = 0
        print(f"Currently comparing {list_of_primes[lp_index]} with {num}")
        while lp_index < len(list_of_primes):
            if num%list_of_primes[lp_index] == 0:
                print(f"Found equal divisibility between {num} and {list_of_primes[lp_index]}")
                print("Incrementing num")
                num += 1
                break
            else:
                print(f"Didnt find divisibility")
                print(f"Checking next prime by incrementing index")
                lp_index +=1
        else:
            print(f"Didn't find divisibility in any previous primes. Therefore found new prime")
            list_of_primes.append(num)
            num += 1

    return list_of_primes[number-1]
```
