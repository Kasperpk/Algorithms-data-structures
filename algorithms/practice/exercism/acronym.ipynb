The cool thing about the findall is that it automatically identifies two words when you have something like 'metal-oxide' and so it does exactly what i've been doing more manually. Below here is an example.


```python
import re

"".join(item[0].upper() for item in re.findall(pattern=r"[a-zA-Z']+", string="metal-oxide"))
```

[a-zA-Z]+: This only looks for pure letters. When it hits the hyphen -, it stops, returns "metal", skips the hyphen, and then finds "oxide".

If i included the '-' in the pattern it would then return it as a single word. 

```python
import re

"".join(item[0].upper() for item in re.findall(pattern=r"[a-zA-Z-]+", string="metal-oxide"))
```

```python
import re
def abbreviate(words):
    return "".join(word[0].upper() for word in re.findall(pattern=r"[a-zA-Z-]+", string="metal-oxide"))
```

```python
import string

def acrynom(text):
    '''
    Function to create acrynom from text input. 
    '''

    split = text.replace('-', ' ').split()

    acrynom = ''

    for word in split:
        cleaned_word = ''
        for letter in word:
            if letter not in string.punctuation:
                cleaned_word = cleaned_word + letter

        acrynom = acrynom + cleaned_word[0].upper()
            

    return acrynom
```
